Extract or filter important data from extracted data 

```python
from langchain_core.documents import Document
```

```python


def filter_data(extracted_data):
    filtered_documents = []

    for doc in extracted_data:
        source = doc.metadata.get("source")
        page = doc.metadata.get("page")
        content = doc.page_content

        filtered_documents.append(
            Document(
                page_content=content,
                metadata={
                    "page": page,
                    "source": source
                }
            )
        )

    return filtered_documents


filtered_data = filter_data(extracted_data)



    
    
```

Chunking Operation 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
def split_text(filtered_doc):
    text_splitting=RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=20,
        length_function=len,
        add_start_index=True
    )
    chunks=text_splitting.split_documents(filtered_doc)
    return chunks

all_chunks=split_text(filtered_data)

print(len(all_chunks))
```

import pinecone

```python
from pinecone import Pinecone

PINECONE_API_KEY = os.getenv("PINECONE_API_KEY")

# Connect to Pinecone
pc = Pinecone(api_key=PINECONE_API_KEY)

print("Pinecone client initialized successfully!")
```

```python
pc
```

Create Index 

```python

from pinecone import ServerlessSpec
index_name = "medicalbot"

# Check whether the index already exists
if index_name not in pc.list_indexes().names():
    pc.create_index(
        name=index_name,
        dimension=768,  # Must match your embedding model
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
    print("Index created successfully!")
else:
    print("Index already exists!")


index = pc.Index(index_name)
print(index.describe_index_stats())
```

Store vectors in the database

```python
from langchain_pinecone import PineconeVectorStore


vector_store = PineconeVectorStore(
    index=index,
    embedding=model,
   
)
```

Add document to vector store

```python

def add_document(doc: Document):
    vector_store = PineconeVectorStore(
        index=index,
        embedding=model
    )

    ids = vector_store.add_documents(
        documents=[doc]
    )

    print("Successfully added extra document!")
    print("Document IDs:", ids)

    return ids

add_document(extra_document)
```

Create Retreiver 

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 4}
)
```

Connect LLM(openAI)

```python
from langchain_openai import ChatOpenAI
```

```python

llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0.6,
    api_key=os.getenv("OPENAI_API_KEY") 
)
```

Generate prompt

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import (
    create_stuff_documents_chain
)


system_prompt=("""
You are a medical information assistant.

Use BOTH:
1. RETRIEVED CONTEXT: information from the user's medical knowledge base.
2. GENERAL KNOWLEDGE: your own learned knowledge.

Instructions:
- First, use the retrieved context when it is relevant.
- You may supplement it with reliable general medical knowledge.
- Do not claim that general knowledge came from the retrieved documents.
- If the sources disagree, explain the disagreement and do not silently
  combine conflicting claims.
- If the retrieved context is insufficient, say so and clearly identify
  any additional general information you provide.
- Never invent citations or claim that a source supports something it does not.
- Do not diagnose the user or prescribe medication.
- If the question describes a possible emergency, prioritise urgent medical help.

RETRIEVED CONTEXT:"""
"{context}"
)

```

```python
prompt=ChatPromptTemplate.from_messages(
    [
("system",system_prompt),
("human","{input}"),

    ]
)
```

```python
prompt
```

```python
document_chain=create_stuff_documents_chain(llm, prompt)
rag_chain=create_retrieval_chain(retriever,document_chain)
```

```python
response=rag_chain.invoke({"input":"what is HIV"})
print(response["answer"])
```
